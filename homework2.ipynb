```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
          "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]
plan = [100, 110, 120, 130, 140, 150, 160, 155, 150, 160, 170, 180]
fact = [95, 115, 118, 125, 150, 145, 170, 150, 140, 165, 175, 190]

plt.figure(figsize=(10, 5))
plt.plot(months, plan, marker="o", label="План")
plt.plot(months, fact, marker="s", label="Факт")
plt.xlabel("Місяць")
plt.ylabel("Продажі, тис. грн")
plt.title("Планові та фактичні продажі за місяцями")
plt.legend()
plt.show()
```

```python
ages = np.random.randint(18, 70, size=100)

plt.figure(figsize=(8, 5))
plt.hist(ages, bins=10, edgecolor="black")
plt.axvline(ages.mean(), color="red", linestyle="--",
            label=f"Середнє: {ages.mean():.1f}")
plt.xlabel("Вік")
plt.ylabel("Кількість осіб")
plt.title("Розподіл віку 100 осіб")
plt.legend()
plt.show()
```

```python
group_a = np.random.normal(75, 8, 30)
group_b = np.random.normal(70, 12, 30)
group_c = np.random.normal(80, 5, 30)

plt.figure(figsize=(8, 5))
plt.boxplot([group_a, group_b, group_c], tick_labels=["Група A", "Група B", "Група C"])
plt.xlabel("Група")
plt.ylabel("Оцінка")
plt.title("Розкид оцінок за групами")
plt.show()
```

```python
dates = ["01.11", "02.11", "03.11", "04.11", "05.11", "06.11", "07.11"]
temperature = [12, 14, 13, 16, 18, 17, 15]
humidity = [70, 65, 72, 60, 55, 58, 63]

plt.figure(figsize=(9, 5))
plt.plot(dates, temperature, marker="o", label="Температура, °C")
plt.plot(dates, humidity, marker="s", label="Вологість, %")
plt.xlabel("Дата")
plt.ylabel("Значення")
plt.title("Температура і вологість за тиждень")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
hours = list(range(24))
load = [20, 15, 12, 10, 10, 15, 25, 40, 60, 75, 80, 85,
        82, 78, 80, 85, 90, 88, 75, 60, 50, 40, 30, 25]

plt.figure(figsize=(10, 5))
plt.plot(hours, load, color="tab:blue", label="Навантаження")
plt.fill_between(hours, load, alpha=0.3)
plt.grid(True)
plt.xlabel("Година доби")
plt.ylabel("Навантаження, %")
plt.title("Навантаження сервера за добу")
plt.xticks(hours)
plt.legend()
plt.show()
```

```python
months = np.arange(1, 13)
conversion = np.random.uniform(2, 6, 12)
retention = np.random.uniform(40, 80, 12)
avg_check = np.random.uniform(500, 1500, 12)
orders = np.random.randint(100, 500, 12)

fig, axs = plt.subplots(2, 2, figsize=(12, 8))

axs[0, 0].plot(months, conversion, marker="o", color="tab:blue")
axs[0, 0].set_title("Конверсія, %")

axs[0, 1].plot(months, retention, marker="o", color="tab:green")
axs[0, 1].set_title("Утримання, %")

axs[1, 0].plot(months, avg_check, marker="o", color="tab:orange")
axs[1, 0].set_title("Середній чек, грн")

axs[1, 1].bar(months, orders, color="tab:red")
axs[1, 1].set_title("Кількість замовлень")

for ax in axs.flat:
    ax.set_xlabel("Місяць")
    ax.grid(True)

fig.suptitle("Метрики продукту за рік", fontsize=14)
plt.tight_layout()
plt.show()
```

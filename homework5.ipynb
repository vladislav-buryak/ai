```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error

screens = np.array([2, 4, 6, 8, 10, 12])
cost = np.array([3.0, 5.5, 7.8, 10.5, 12.9, 15.8])  # тис. $

df = pd.DataFrame({"screens": screens, "cost": cost})
print(df)
```

```python
X = df[["screens"]]   # scikit-learn очікує 2D-масив ознак
y = df["cost"]

model = LinearRegression()
model.fit(X, y)

print(f"Коефіцієнт (вартість одного екрана): {model.coef_[0]:.3f} тис. $")
print(f"Вільний член (базова вартість): {model.intercept_:.3f} тис. $")

y_pred = model.predict(X)
print(f"R²: {r2_score(y, y_pred):.4f}")
print(f"MAE: {mean_absolute_error(y, y_pred):.3f} тис. $")
```

```python
new_projects = pd.DataFrame({"screens": [5, 9, 15]})
predictions = model.predict(new_projects)

for s, p in zip(new_projects["screens"], predictions):
    print(f"{s} екранів → приблизно {p:.1f} тис. $")
```

```python
line_x = pd.DataFrame({"screens": np.linspace(0, 16, 100)})

plt.figure(figsize=(8, 5))
plt.scatter(df["screens"], df["cost"], color="tab:blue", label="Історичні замовлення")
plt.plot(line_x["screens"], model.predict(line_x), color="tab:red", label="Лінія регресії")
plt.scatter(new_projects["screens"], predictions, color="tab:green",
            marker="*", s=150, label="Прогноз для нових клієнтів")
plt.xlabel("Кількість екранів")
plt.ylabel("Вартість, тис. $")
plt.title("Залежність вартості розробки від кількості екранів")
plt.legend()
plt.grid(True)
plt.show()
```

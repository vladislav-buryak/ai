```python
import numpy as np
```

```python
arr = np.random.randint(0, 101, size=10)

print("Масив до сортування:", arr)
print("Мінімум:", np.min(arr))
print("Максимум:", np.max(arr))
print("Середнє:", np.mean(arr))

sorted_arr = np.sort(arr)
print("Масив після сортування:", sorted_arr)
```

```python
matrix = np.random.randint(1, 51, size=(4, 4))

print("Матриця 4x4:")
for row in matrix:
    print(" ".join(f"{x:3d}" for x in row))

print("Суми за рядками:  ", matrix.sum(axis=1))
print("Суми за стовпцями:", matrix.sum(axis=0))
print("Загальне середнє: ", matrix.mean())
```

```python
a = np.arange(15).reshape(3, 5)
print("Вихідний масив 3x5:")
print(a)

t = a.T
print("\nТранспонований масив 5x3:")
print(t)

r = a.reshape(5, 3)
print("\nReshape до 5x3:")
print(r)
```

```python
m = np.random.rand(10, 10)
print("Масив 10x10:")
print(m)

min_val, max_val = m.min(), m.max()
min_idx = np.unravel_index(np.argmin(m), m.shape)
max_idx = np.unravel_index(np.argmax(m), m.shape)

print(f"\nМінімум: {min_val:.4f}, індекс (рядок, стовпець): {min_idx}")
print(f"Максимум: {max_val:.4f}, індекс (рядок, стовпець): {max_idx}")
```

```python
arr = np.random.randint(-10, 11, size=15)
print("Масив:", arr)

positive = arr[arr > 0]
even = arr[arr % 2 == 0]

print("Додатні елементи:", positive)
print("Парні елементи:  ", even)
```

```python
a = np.arange(1, 6)    # [1 2 3 4 5]
b = np.arange(6, 11)   # [6 7 8 9 10]

# 1. Конкатенація в один масив
concat = np.concatenate([a, b])
print("Конкатенація:", concat)

# 2. Двовимірний масив 2x5
two_rows = np.vstack([a, b])      # або np.array([a, b])
print("\nМасив 2x5:")
print(two_rows)

# 3. Об'єднання в колонки 5x2
two_cols = np.column_stack([a, b])  # або np.stack([a, b], axis=1)
print("\nМасив 5x2:")
print(two_cols)
```

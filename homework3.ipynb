```python
import numpy as np
```

```python
image_matrix = np.array([
    [100, 150, 200, 250],
    [50,  120, 180, 220],
    [30,   90, 140, 190],
    [10,   50, 100, 150]
])

darker = image_matrix - 20
darker = np.clip(darker, 0, 255)

print("Затемнене зображення:")
print(darker)
print("Середня інтенсивність:", darker.mean())
```

```python
coordinates = np.array([[1.5, 2.0], [5.0, 8.1], [3.2, 1.1], [8.9, 4.3], [2.1, 2.8]])

x, y = coordinates.T

print("X:", x)
print("Y:", y)
print("Максимум X:", x.max())
print("Мінімум Y:", y.min())

mask = (x + y) > 6.0
print("Точки з X + Y > 6.0:")
print(coordinates[mask])
```

```python
stock_prices = np.array([100.0, 102.5, 101.0, 105.0, 104.2, 108.0, 110.5])

today = stock_prices[1:]
yesterday = stock_prices[:-1]

returns = (today - yesterday) / yesterday
print("Дохідність за день:", np.round(returns, 4))

best = np.argmax(returns)
print(f"Найвища дохідність: {returns[best]:.4%}")
print("Індекс у масиві returns:", best)
print("Це день №", best + 2, "від початку періоду (нумерація з 1)")
```

```python
sensor_data = np.array([23.4, 24.1, -999, 25.0, 26.2, -999, 24.8])

mask = sensor_data == -999
print("Маска пропусків:", mask)

mean_valid = sensor_data[~mask].mean()
print("Середнє валідних значень:", mean_valid)

filled = np.where(mask, mean_valid, sensor_data)
print("Масив після заміни:", filled)
```

```python
inventory = np.array([12, 45, 23, 67, 89, 34, 50])
target = 40

diff = np.abs(inventory - target)
print("Абсолютні різниці:", diff)

idx = np.argmin(diff)
print("Індекс найближчого:", idx)
print("Найближче значення:", inventory[idx])
```
